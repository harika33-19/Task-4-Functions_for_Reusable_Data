```python
def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"
```

```python
df["Total"] = df[["Maths", "Science", "English"]].sum(axis=1)

df["Mean"] = df[["Maths", "Science", "English"]].mean(axis=1)

df["Percentage"] = df["Total"].apply(
    lambda x: calculate_percentage([x], 300)
)

df["Grade"] = df["Percentage"].apply(calculate_grade)

df
```

```python
def calculate_mean(values):
    return sum(values) / len(values)


def find_maximum(values):
    return max(values)


def find_minimum(values):
    return min(values)


def calculate_percentage(marks, total_marks):
    return (sum(marks) / total_marks) * 100


def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"
```

```python
marks = [85, 90, 80]

mean = calculate_mean(marks)
maximum = find_maximum(marks)
minimum = find_minimum(marks)
percentage = calculate_percentage(marks, 300)
grade = calculate_grade(percentage)

print("Marks:", marks)
print("Mean:", mean)
print("Maximum:", maximum)
print("Minimum:", minimum)
print("Percentage:", percentage)
print("Grade:", grade)
```

```python
# Functions for Reusable Data Tasks

## Objective
To create reusable Python functions for calculations, data transformation, and simple statistical operations.

## Functions Used

1. calculate_mean() - Calculates the mean of values.
2. find_maximum() - Finds the maximum value.
3. find_minimum() - Finds the minimum value.
4. calculate_percentage() - Calculates percentage.
5. calculate_grade() - Assigns a grade based on percentage.

## Dataset
Student Marks Dataset containing Maths, Science and English marks.

## Conclusion
Functions make Python programs more organized, reusable and readable.
Using parameters and return values allows the same logic to be reused for different data.
```
